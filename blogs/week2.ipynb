## Day 1: Laying the Foundation

On the first day, my main goal was to **structure the project** in a way that would be scalable and maintainable. I created three main backend modules:
1. **data_loader** – for loading datasets efficiently.
2. **EDAEngine** – for exploratory data analysis functions like summary statistics, missing value checks, and feature insights.
3. **VizEngine** – for generating visualizations like histograms, scatter plots, and correlation matrices.

I learned that even small projects benefit from **system design principles**, like separation of concerns and modular coding. Initially, I tried putting all functions in a single file, which quickly became messy and hard to debug. Refactoring into separate modules taught me a lot about maintainable code.

```python
# main.py - Initial setup
from app_backend import data_loader, EDAEngine, VizEngine

# Load dataset
data = data_loader.load_csv('data/sales_data.csv')

# Initialize EDA Engine
eda = EDAEngine(data)

# Initialize Visualization Engine
viz = VizEngine(data)

# Run EDA
eda.summary()  # Overview of dataset
eda.missing_values()  # Check missing data

# Quick visualizations
viz.histogram('Revenue')
viz.scatter('Revenue', 'Profit')
```

### Lessons Learned
- **Importance of modularity:** Refactoring improved readability and reduced bugs.
- **Debugging skills:** Encountered issues with missing values and type mismatches.
- **System design basics:** Learned to plan module interactions and maintain separation of concerns.

## Day 2: Streamlit Integration & Interactive Dashboards

Day 2 focused on **building the frontend using Streamlit**. I wanted to make the project interactive, allowing users to explore correlations and visualizations dynamically. I spent some time reading **"Hands-On Streamlit for Data Science" by Jake VanderPlas** (just to give context) and experimenting with widgets, layout, and markdown styling.

Challenges I faced included:
- **Widget dependencies:** Initially the dropdown selection caused crashes due to incompatible column types.
- **Large dataset rendering:** Plots lagged when the dataset was too large, which taught me optimization using sampling.
- **Correlation heatmap clarity:** Needed color scaling and annotations to make insights readable.

I overcame these issues by writing utility functions in `VizEngine` that handle data types dynamically and by implementing caching using `@st.cache_data` for large datasets.

```python
# Streamlit Page - 3_Correlation.py
import streamlit as st
from app_backend import data_loader, EDAEngine, VizEngine

# Page config
st.set_page_config(page_title="Correlation Analysis", page_icon="📉", layout="wide")

st.title("📉 Correlation Analysis")
st.markdown("Explore numeric feature correlations interactively.")

# Load data
data = data_loader.load_csv('data/sales_data.csv')

# Initialize engines
eda = EDAEngine(data)
viz = VizEngine(data)

# Correlation matrix
st.write(data.corr())

# Interactive scatter plot
numeric_cols = data.select_dtypes(include='float64').columns
x_feature = st.selectbox('Select X-axis', numeric_cols)
y_feature = st.selectbox('Select Y-axis', numeric_cols)

# Display plot
st.write(viz.scatter(x_feature, y_feature))
```

### Reflections & Overcoming Mistakes
- Learned the importance of **caching data** for performance.
- Realized that **modular visualization functions** reduce repetition and bugs.
